```python
from huggingface_hub import HfApi

api = HfApi()
# Identificador del repositorio de modelos
model_repo_id = "victortomeno/iris26"

# Creacion del repositorio de tipo modelo
api.create_repo(repo_id=model_repo_id, repo_type="model", exist_ok=True)
```

```python
# Subida del modelo Decision Tree
api.upload_file(
    path_or_fileobj="models/iris_dt.joblib",
    path_in_repo="iris_dt.joblib",
    repo_id=model_repo_id,
    repo_type="model"
)

# Subida del modelo Logistic Regression
api.upload_file(
    path_or_fileobj="models/iris_logreg.joblib",
    path_in_repo="iris_logreg.joblib",
    repo_id=model_repo_id,
    repo_type="model"
)
```

```python
# Contenido basico de la Model Card
content = """
# Iris Classification Models
Modelos entrenados para la clasificacion del dataset Iris:
- Decision Tree
- Logistic Regression
"""

with open("README.md", "w") as f:
    f.write(content)

# Subida del archivo README al repositorio
api.upload_file(
    path_or_fileobj="README.md",
    path_in_repo="README.md",
    repo_id=model_repo_id,
    repo_type="model"
)
```
